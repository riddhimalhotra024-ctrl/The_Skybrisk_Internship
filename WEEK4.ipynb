```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Line Plot
# -------------------------------
months = ['Jan', 'Feb', 'Mar', 'Apr']
sales = [100, 120, 150, 170]

plt.figure(figsize=(6,4))
plt.plot(months, sales, marker='o')
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
# Histogram
# -------------------------------
marks = [65, 70, 75, 80, 85, 90, 95, 80, 75]

plt.figure(figsize=(6,4))
plt.hist(marks, bins=7)
plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
# Heatmap
# -------------------------------
data = pd.DataFrame({
    'Math': [80, 85, 90, 95],
    'Science': [75, 80, 85, 90],
    'English': [70, 75, 80, 85]
})

plt.figure(figsize=(6,4))
sns.heatmap(data.corr(), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

```python
# Dataset Dashboard
# -------------------------------
df = pd.DataFrame({
    'Study_Hours': [2, 3, 5, 4, 1],
    'Marks': [80, 75, 90, 85, 70],
    'Attendance': [85, 88, 95, 90, 80]
})

# Scatter Plot
plt.figure(figsize=(6,4))
sns.scatterplot(x='Study_Hours', y='Marks', data=df)
plt.title("Study Hours vs Marks")
plt.show()

# Histogram
plt.figure(figsize=(6,4))
plt.hist(df['Marks'], bins=10)
plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()

# Pairplot
sns.pairplot(df)
plt.show()
```
