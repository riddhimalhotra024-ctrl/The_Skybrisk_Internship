NumPy and Pandas for Data Manipulation

```python

import numpy as np
import pandas as pd
```

```python
# NumPy Operations
# -------------------------------
arr = np.array([10, 20, 30, 40, 50])

print("=== NumPy Operations ===")
print("Array:", arr)
print("Mean:", np.mean(arr))
print("Sum:", np.sum(arr))
print("Square of Elements:", arr**2)
```

```python
# Pandas DataFrame Creation
# -------------------------------
print("\n=== Student DataFrame ===")

student_data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35]
}

students_df = pd.DataFrame(student_data)

print(students_df)
```

```python
# Dataset Cleaning and Aggregation
# -------------------------------
print("\n=== Dataset Cleaning ===")

marks_data = {
    'Name': ['A', 'B', 'C', 'D'],
    'Marks': [80, None, 90, 85]
}

marks_df = pd.DataFrame(marks_data)

print("Original Dataset:")
print(marks_df)

# Remove missing values
clean_df = marks_df.dropna()

print("\nCleaned Dataset:")
print(clean_df)

# Calculate average marks
average_marks = clean_df['Marks'].mean()

print("\nAverage Marks:", average_marks)
```

```python
# Grouping Data
# -------------------------------
print("\n=== Grouping Data ===")

salary_data = {
    'Department': ['IT', 'HR', 'IT', 'HR', 'Finance'],
    'Salary': [50000, 40000, 55000, 42000, 60000]
}

salary_df = pd.DataFrame(salary_data)

grouped_salary = salary_df.groupby('Department')['Salary'].mean()

print("Average Salary by Department:")
print(grouped_salary)
```

```python
##EMPLOYEE DATASET ANALYSIS
import pandas as pd

data = {
    "Employee": [
        "A",
        "B",
        "C",
        "D"
    ],

    "Salary": [
        50000,
        60000,
        55000,
        70000
    ]
}

df = pd.DataFrame(data)

print("Average Salary:")
print(df["Salary"].mean())

print("Maximum Salary:")
print(df["Salary"].max())
```

```python
##DATA CLEANING AND AGGREGATION
import pandas as pd
import numpy as np

data = {
    "Name": [
        "A",
        "B",
        "C",
        "D",
        "E"
    ],

    "Marks": [
        80,
        np.nan,
        90,
        85,
        np.nan
    ]
}

df = pd.DataFrame(data)

print("Original Dataset:\n")
print(df)

average_marks = df["Marks"].mean()

df["Marks"].fillna(
    average_marks,
    inplace=True
)

print("\nCleaned Dataset:\n")
print(df)

print(
    "\nAverage Marks:",
    df["Marks"].mean()
)
```
