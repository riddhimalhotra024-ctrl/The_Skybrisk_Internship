List

```python
##A list is an ordered collection of items. Lists are mutable, meaning their contents can be modified after creation.
lst=["apple","banana","mango"]
print(lst)
print(lst[0])
print(lst[2])
```

Tuple

```python
##A tuple is similar to a list but cannot be modified after creation. Tuples are immutable.
tpl=("apple",1,"hello")
print(tpl)
print(tpl[0])
print(tpl[1])
print(tpl[2])
```

Dictionary

```python
##A dictionary stores data in key-value pairs.
student = {
    "name": "Riddhi",
    "age": 21,
    "city": "Varanasi"
}

print(student["name"])
print(student["age"])
print(student["city"])
```

Sets

```python
##A set stores unique values only.
num={1,2,3,4,5,6,4,3,2,1,3,2,3,5}
print(num)
```

Functions

```python
##Functions are reusable blocks of code that perform a specific task.
def greet():
    print("hello world!")
greet()    
```

Lambda Function

```python
##Lambda functions are small anonymous functions written in a single line.
square = lambda x:x*x
print(square(5))
```

Recursion


```python
##Recursion occurs when a function calls itself.
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```

List Comprehension

```python
##List comprehension provides a concise way to create lists.
squares = [i * i for i in range(5)]

print(squares)
```

SUM OF SQAURES

```python
numbers = [1, 2, 3, 4, 5]

def sum_of_squares(data):
    return sum(x**2 for x in data)

result = sum_of_squares(numbers)

print("Sum of Squares:", result)
```

FILTERING

```python
numbers = [10, 15, 22, 33, 40, 55]

def filter_even(data):
    return [num for num in data if num % 2 == 0]

print(filter_even(numbers))
```

DATA CLEANING

```python
data = [10, 20, 30, 20, 40, 50, 10, 60]

unique_data = list(set(data))

filtered_data = [x for x in unique_data if x > 25]

print("Unique Data:", unique_data)
print("Filtered Data:", filtered_data)
```
